# DataLoader Test Notebook

`sw-framework-v001`의 `DataModule`을 빠르게 점검하는 테스트 노트북입니다.

- CSV 로드
- train/val/test 분할
- DataLoader 배치 shape 확인

```python
from pathlib import Path
import sys
from types import SimpleNamespace

# 프로젝트 루트/소스 경로 등록
project_root = Path('/data2/wk/SW_framework/workdir/sw-framework-v001')
src_root = project_root / 'src'
if str(src_root) not in sys.path:
    sys.path.insert(0, str(src_root))

from data.loader import DataModule
```

```python
# 테스트용 설정
cfg = SimpleNamespace(
    data_path='/data2/wk/SW_framework/data/xray_hourly.csv',
    target_col='Long',
    feature_cols=['Year', 'Month', 'Day', 'Hour'],
    time_col=None,
    seq_len=24,
    pred_len=1,
    train_ratio=0.7,
    val_ratio=0.15,
    batch_size=32,
    shuffle_train=True,
    num_workers=0,
)

dm = DataModule(cfg)
bundle = dm.setup()

print('input_size:', bundle.input_size)
print('target_index:', bundle.target_index)
print('train batches:', len(bundle.train_loader))
print('val batches:', len(bundle.val_loader))
print('test batches:', len(bundle.test_loader))
```

```python
# 배치 shape 확인
batch_x, batch_y = next(iter(bundle.train_loader))
print('batch_x shape:', tuple(batch_x.shape))  # (B, seq_len, num_features)
print('batch_y shape:', tuple(batch_y.shape))  # (B, pred_len, 1)

# 값 샘플 확인
print('batch_x[0, :3, :]:\n', batch_x[0, :3, :])
print('batch_y[0]:\n', batch_y[0])
```
